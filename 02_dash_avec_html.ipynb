# Utilisation de dash avec HTML

## Ajout d'une illustation au code précedent ( avec des styles)

```python
import dash
from dash import html, dcc, Input, Output, callback
import plotly.graph_objects as go

app = dash.Dash()

app.layout = html.Div(children=[
    html.H2(children="Hello World!", style={'color':'red'}),
    html.H2(children="Une illustration", style={'color':'blue'}),

    dcc.Graph(
        figure= go.Figure([
            go.Bar(
                x=[1,2,3],
                y=[2,1,3]
            )
        ])

        )

])

@callback(
    Output(component_id='dic-1', component_property='children'),
    [Input(component_id='input-1', component_property='value')]
)
def uptdate_div(input):
    return f'Texte introduit : {input}'

if __name__ == "__main__":
    app.run(debug=False, port=8050, jupyter_mode="external")
```

```python
import dash
from dash import html, dcc, Input, Output, callback
import plotly.graph_objects as go
import pandas_datareader.data as web

app = dash.Dash()

df = web.DataReader(name="AMZN", data_source='stooq')
df.reset_index(inplace=True)
df = df[:30]
min_val = min(len(df), 30)
app.layout = html.Div(children=[
    html.H2(children="Hello World!", style={'color':'red'}),

    html.Table([
        html.Tr([html.Th(col) for col in df.columns])] +
        [html.Tr([html.Td(df.iloc[i][col]) for col in df.columns]) for i in range(min_val)]
    )

])

@callback(
    Output(component_id='dic-1', component_property='children'),
    [Input(component_id='input-1', component_property='value')]
)
def uptdate_div(input):
    return f'Texte introduit : {input}'

if __name__ == "__main__":
    app.run(debug=False, port=8050, jupyter_mode="external")
```
